```python
import pandas as pd
import matplotlib.pyplot as plt
```

### 1. Kurtosis ###

```python
df = pd.read_excel("listings_adjusted_iqr_bereinigt.xlsx")
kurtosis_wert = df['review_scores_rating'].kurt()

print(f"Die Kurtosis beträgt: {kurtosis_wert:.4f}")
```

Die Verteilung von review_scores_rating ist stark spitzgipflig bzw. hat sehr ausgeprägte Ausreißer-/Randbereiche im Vergleich zur Normalverteilung.

### 2. Schiefe ###

```python
schiefe_wert = df['review_scores_rating'].skew()

print(f"Die Schiefe beträgt: {schiefe_wert:.4f}")
```

Ein Wert im negativen Bereich zeigt an, dass die Daten linksschief sind (es existiert somit ein längerer linker Rand)

```python
import pandas as pd
import numpy as np

spalte = 'review_scores_rating'

werte = df[spalte].dropna()
n_gesamt = len(werte)

ergebnisse = []

# Schrittweite: bei Ratings mit zwei Nachkommastellen ist 0.01 sinnvoll
untere_grenzen = np.arange(werte.min(), werte.max(), 0.01)
obere_grenzen = np.arange(werte.min(), werte.max(), 0.01)

# Gewichtung
gewicht_schiefe = 1
gewicht_kurtosis = 1
gewicht_anzahl = 1
gewicht_spannweite = 1

for untergrenze in untere_grenzen:
    for obergrenze in obere_grenzen:
        if untergrenze < obergrenze:
            
            gefiltert = werte[
                (werte >= untergrenze) &
                (werte <= obergrenze)
            ]
            
            # Mindestanzahl, damit keine Mini-Bereiche gewinnen
            if len(gefiltert) >= 100:
                schiefe = gefiltert.skew()
                kurtosis = gefiltert.kurt()
                
                anzahl = len(gefiltert)
                anteil = anzahl / n_gesamt
                spannweite = obergrenze - untergrenze
                relative_spannweite = spannweite / (werte.max() - werte.min())
                
                # Je kleiner, desto besser
                normalitaets_abstand = (
                    gewicht_schiefe * abs(schiefe) +
                    gewicht_kurtosis * abs(kurtosis)
                )
                
                # Je größer Anteil und Spannweite, desto besser
                # Deshalb werden sie abgezogen
                score = (
                    normalitaets_abstand
                    - gewicht_anzahl * anteil
                    - gewicht_spannweite * relative_spannweite
                )
                
                ergebnisse.append({
                    'untergrenze': round(untergrenze, 2),
                    'obergrenze': round(obergrenze, 2),
                    'anzahl_werte': anzahl,
                    'anteil_in_prozent': round(anteil * 100, 2),
                    'spannweite': round(spannweite, 2),
                    'schiefe': round(schiefe, 4),
                    'kurtosis': round(kurtosis, 4),
                    'normalitaets_abstand': round(normalitaets_abstand, 4),
                    'score': round(score, 4)
                })

ergebnisse_df = pd.DataFrame(ergebnisse)

beste_bereiche = ergebnisse_df.sort_values('score').head(20)

print(beste_bereiche.to_string(index=False))
```
