```python
import pandas as pd

def populationsanalyse(csv_datei, output_datei="populationsanalyse.xlsx"):

    # CSV laden
    df = pd.read_csv(csv_datei)

    # Spalten ausschließen
    ausgeschlossene_spalten = [
        "id",
        "listing_url"
    ]

    # Excel-Datei erstellen
    with pd.ExcelWriter(output_datei, engine="openpyxl") as writer:

        for spalte in df.columns:

            # Überspringen
            if spalte in ausgeschlossene_spalten:
                continue

            serie = df[spalte]

            print(f"\n=== Analyse für: {spalte} ===")

            # Numerische Spalten -> Quartile
            if pd.api.types.is_numeric_dtype(serie):

                klassen = pd.qcut(
                    serie,
                    q=4,
                    duplicates="drop"
                )

                tabelle = (
                    klassen.value_counts()
                    .sort_index()
                    .reset_index()
                )

                tabelle.columns = ["Quartil", "Anzahl"]

            # Kategoriale/Text-Spalten
            else:

                tabelle = (
                    serie.value_counts()
                    .reset_index()
                )

                tabelle.columns = ["Klasse", "Anzahl"]

            # Prozent ergänzen
            tabelle["Prozent"] = round(
                tabelle["Anzahl"] / len(df) * 100,
                2
            )

            # Anzeigen
            print(tabelle)

            # In Excel speichern
            tabelle.to_excel(
                writer,
                sheet_name=str(spalte)[:31],
                index=False
            )

    print(f"\nAnalyse gespeichert als: {output_datei}")
```

```python
populationsanalyse("listings_clean.csv")
```
