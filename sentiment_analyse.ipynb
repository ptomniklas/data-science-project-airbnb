```python
import pandas as pd

reviews = pd.read_excel("reviews_neu.xlsx")
listings = pd.read_csv("listings_clean.csv")

print("Reviews-Datei:")
print(reviews.shape)
print(reviews.columns.tolist())

print("\nListings-Datei:")
print(listings.shape)
print(listings.columns.tolist())
```

```python
import pandas as pd

reviews_text = pd.read_csv("reviews.csv.gz", compression="gzip")

print("Reviews-Text-Datei:")
print(reviews_text.shape)
print(reviews_text.columns.tolist())

reviews_text.head()
```

```python
# Listings-Datei laden
listings = pd.read_csv("listings_clean.csv")

# Review-Texte mit den Listing-Bewertungen verbinden
df = reviews_text.merge(
    listings[["id", "review_scores_rating"]],
    left_on="listing_id",
    right_on="id",
    how="left"
)

print(df.shape)
df.head()
```

```python
# Bereinigung der Daten

#  Zeilen ohne Kommentar oder ohne Rating entfernen
df = df.dropna(subset=["comments", "review_scores_rating"])

# Kommentare als Text behandeln
df["comments"] = df["comments"].astype(str)

# Sehr kurze Kommentare entfernen
df = df[df["comments"].str.len() > 10]

print(df.shape)

df[["listing_id", "comments", "review_scores_rating"]].head(10)
```

```python
# Erstellen eines Testdatensatzes mit 100 Reviews
df_test = df.head(100).copy()

print(df_test.shape)

df_test[["listing_id", "comments", "review_scores_rating"]].head(10)
```

```python
# Laden des Modells
from transformers import pipeline

sentiment_model = pipeline(
    "sentiment-analysis",
    model="nlptown/bert-base-multilingual-uncased-sentiment"
)

print("Modell geladen.")
```

```python
# Modell testen
test_texts = [
    "Excellent stay, I would highly recommend it.",
    "The apartment was okay, but a bit noisy.",
    "The room was dirty and the host was unfriendly."
]

for text in test_texts:
    result = sentiment_model(text)[0]
    print(text)
    print(result)
    print()
```

```python
# Funktion bauen
def predict_sentiment_stars(text):
    result = sentiment_model(str(text)[:512])[0]
    label = result["label"]  # z.B. "5 stars" oder "1 star"
    stars = int(label[0])
    return stars
```

```python
df_test["predicted_sentiment_stars"] = df_test["comments"].apply(predict_sentiment_stars)

df_test[["comments", "review_scores_rating", "predicted_sentiment_stars"]].head(20)
```

```python
# Differenz zwischen tatsächlicher Bewertung und Modellvorhersage.
df_test["difference"] = df_test["review_scores_rating"] - df_test["predicted_sentiment_stars"]

df_test[["comments", "review_scores_rating", "predicted_sentiment_stars", "difference"]].head(20)
```

```python
# Korrelation berechnen
correlation = df_test["review_scores_rating"].corr(df_test["predicted_sentiment_stars"])

print("Korrelation:", correlation)
```

```python
# Zufällige Stichprobe von 100 Reviews ziehen.
# df.head(100) nimmt nur die ersten 100 Zeilen. Diese können alle zum gleichen Listing gehören,
# wodurch die echten Bewertungen kaum variieren. Eine Zufallsstichprobe ist deshalb sinnvoller.
df_test_random = df.sample(100, random_state=42).copy()

# Für jeden Review-Text wird mit dem vortrainierten Sentiment-Modell
# eine geschätzte Sternebewertung von 1 bis 5 erzeugt.
df_test_random["predicted_sentiment_stars"] = df_test_random["comments"].apply(
    predict_sentiment_stars
)

# Differenz zwischen echter Airbnb-Bewertung und modellbasierter Textbewertung berechnen.
# Positive Werte bedeuten: Die echte Bewertung ist höher als die Stimmung im Text.
# Negative Werte bedeuten: Die Textbewertung fällt höher aus als die echte Airbnb-Bewertung.
df_test_random["difference"] = (
    df_test_random["review_scores_rating"] 
    - df_test_random["predicted_sentiment_stars"]
)

# Ergebnisübersicht anzeigen:
# comments = Bewertungstext
# review_scores_rating = echte Airbnb-Bewertung des Listings
# predicted_sentiment_stars = vom Modell geschätzte Bewertung aus dem Text
# difference = Abweichung zwischen echter Bewertung und Text-Sentiment
df_test_random[
    ["comments", "review_scores_rating", "predicted_sentiment_stars", "difference"]
].head(20)
```

```python
# Zusammenhang zwischen echter Airbnb-Bewertung und modellbasierter Textbewertung berechnen
correlation = df_test_random["review_scores_rating"].corr(
    df_test_random["predicted_sentiment_stars"]
)

print("Korrelation zwischen echter Bewertung und Text-Sentiment:", correlation)
```

```python
# Durchschnittliche Abweichung zwischen echter Bewertung und Text-Sentiment
mean_difference = df_test_random["difference"].mean()

print("Durchschnittliche Abweichung:", mean_difference)
```

```python
# Häufigkeit der vom Modell vorhergesagten Sternebewertungen
df_test_random["predicted_sentiment_stars"].value_counts().sort_index()
```

```python
# Größere zufällige Stichprobe für die finale Analyse
# 100 Reviews waren nur ein Funktionstest.
# Mit 1.000 Reviews werden die Ergebnisse stabiler und aussagekräftiger.

df_sample = df.sample(1000, random_state=42).copy()

print(df_sample.shape)
df_sample[["listing_id", "comments", "review_scores_rating"]].head(10)
```

```python
# Für jeden Review-Text wird eine Sentiment-Sternebewertung von 1 bis 5 berechnet.

df_sample["predicted_sentiment_stars"] = df_sample["comments"].apply(
    predict_sentiment_stars
)

df_sample[["comments", "review_scores_rating", "predicted_sentiment_stars"]].head(10)
```

```python
# Differenz zwischen echter Airbnb-Bewertung und modellbasierter Textbewertung berechnen.
# Positive Werte: echte Bewertung ist höher als die Textbewertung.
# Negative Werte: Textbewertung ist höher als die echte Bewertung.

df_sample["difference"] = (
    df_sample["review_scores_rating"] 
    - df_sample["predicted_sentiment_stars"]
)

df_sample[
    ["comments", "review_scores_rating", "predicted_sentiment_stars", "difference"]
].head(20)
```

```python
# Korrelation zwischen echter Airbnb-Bewertung und modellbasierter Textbewertung

correlation = df_sample["review_scores_rating"].corr(
    df_sample["predicted_sentiment_stars"]
)

print("Korrelation:", correlation)
```

```python
# Durchschnittliche Abweichung zwischen echter Bewertung und Textbewertung

mean_difference = df_sample["difference"].mean()

print("Durchschnittliche Abweichung:", mean_difference)
```

```python
# Häufigkeiten der vom Modell vorhergesagten Sternebewertungen

df_sample["predicted_sentiment_stars"].value_counts().sort_index()
```

```python
# Zufällige Stichprobe von 50.000 Reviews ziehen
# random_state=42 sorgt dafür, dass die Stichprobe reproduzierbar ist.

df_sample_50000 = df.sample(50000, random_state=42).copy()

print(df_sample_50000.shape)
df_sample_50000[["listing_id", "comments", "review_scores_rating"]].head(10)
```

```python
# 50.000 blockweise analysieren und speichern.

import os
import pandas as pd

# Einstellungen
chunk_size = 10000
output_file = "sentiment_ergebnisse_50000.csv"

# Falls die Datei schon existiert, abbrechen, damit nichts versehentlich doppelt/überschrieben wird.
if os.path.exists(output_file):
    print(f"Achtung: Die Datei '{output_file}' existiert bereits.")
    print("Bitte lösche oder benenne sie um, bevor du die Analyse neu startest.")
else:
    total_rows = len(df_sample_50000)
    print("Anzahl zu analysierender Reviews:", total_rows)

    for start in range(0, total_rows, chunk_size):
        end = min(start + chunk_size, total_rows)
        print(f"Verarbeite Zeilen {start} bis {end} von {total_rows}")

        # Aktuellen Block kopieren
        df_chunk = df_sample_50000.iloc[start:end].copy()

        # Sentiment-Sterne mit dem vortrainierten Modell berechnen
        df_chunk["predicted_sentiment_stars"] = df_chunk["comments"].apply(
            predict_sentiment_stars
        )

        # Differenz zwischen echter Bewertung und Textbewertung berechnen
        df_chunk["difference"] = (
            df_chunk["review_scores_rating"]
            - df_chunk["predicted_sentiment_stars"]
        )

        # Ersten Block mit Spaltennamen speichern, danach an dieselbe Datei anhängen
        write_header = start == 0

        df_chunk.to_csv(
            output_file,
            mode="a",
            header=write_header,
            index=False
        )

        print(f"Gespeichert: Zeilen {start} bis {end}")

    print("Fertig. Alle 50.000 Ergebnisse wurden gespeichert.")
```
